# Comparativa de Modelos — Diabetes SVM

Heatmap de matrices de confusión y gráfica de barras comparativa entre:
- **Grid Search + Threshold Óptimo** (CON HbA1c/glucosa, optimizado)
- **Sin HbA1c/Glucosa** (baseline)

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
modelos = ['Grid Search + Threshold Óptimo', 'Sin HbA1c/Glucosa']

cm = [
    np.array([[17409, 881], [294, 1403]]),
    np.array([[13192, 5098], [328, 1369]])
]

metricas = ['Precision', 'Recall', 'F1']
valores = np.array([[0.614, 0.827, 0.705], [0.212, 0.807, 0.335]])
```

```python
fig = plt.figure(figsize=(16, 6))
gs = fig.add_gridspec(1, 3, width_ratios=[1, 1, 1.3])

for i, (matriz, titulo) in enumerate(zip(cm, modelos)):
    ax = fig.add_subplot(gs[0, i])
    sns.heatmap(matriz, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['No Diabetes', 'Diabetes'],
                yticklabels=['No Diabetes', 'Diabetes'],
                cbar=False, annot_kws={'size': 14})
    ax.set_title(titulo, fontsize=12, fontweight='bold')
    ax.set_ylabel('Real', fontsize=11)
    ax.set_xlabel('Predicho', fontsize=11)
    ax.tick_params(axis='both', labelsize=10)

ax = fig.add_subplot(gs[0, 2])
x = np.arange(len(modelos))
w = 0.25
colores = ['#2ecc71', '#e74c3c', '#3498db']

for j, met in enumerate(metricas):
    bars = ax.bar(x + j * w - w, valores[:, j], w, label=met, color=colores[j], alpha=0.8)
    for bar, v in zip(bars, valores[:, j]):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.02,
                f'{v:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

ax.set_xticks(x)
ax.set_xticklabels(modelos, fontsize=10)
ax.set_ylabel('Score', fontsize=11)
ax.set_ylim([0, 1.1])
ax.set_title('Métricas', fontsize=12, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
